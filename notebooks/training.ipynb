# Training Notebook

This notebook orchestrates the training pipeline using the functions defined in the `src` package. It loads the processed data, builds an InceptionV3‑based classifier, trains it, evaluates it and optionally plots the learning curves.

```python
%load_ext autoreload
%autoreload 2
```

```python
import numpy as np
from pathlib import Path
from src import build_inceptionv3, train_model, evaluate_model, plot_learning_curves
from src.config import PROCESSED_DATA_DIR, MODELS_DIR, FINAL_MODEL_PATH
```

```python
# Load processed numpy arrays
data_path = PROCESSED_DATA_DIR
x_train = np.load(data_path / 'x_train.npy')
y_train = np.load(data_path / 'y_train.npy')
x_test = np.load(data_path / 'x_test.npy')
y_test = np.load(data_path / 'y_test.npy')
print('Data shapes:', x_train.shape, y_train.shape, x_test.shape, y_test.shape)
```

```python
# Build the InceptionV3 model
model = build_inceptionv3()
```

```python
# Train the model
train_result = train_model(model, x_train, y_train, x_test, y_test)
```

```python
# Evaluate on the test set
eval_result = evaluate_model(model, x_test, y_test)
```

```python
# Plot learning curves (if history is available)
if 'history' in train_result:
    plot_learning_curves(train_result['history'])
```
